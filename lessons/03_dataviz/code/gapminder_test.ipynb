```python
import pandas as pd
import pathlib, glob
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
gapminder_csvpath = Path("../../../data/gapminder.csv")
gapminder_df = pd.read_csv(gapminder_csvpath)
```

Seaborn [scatterplot documentation](https://seaborn.pydata.org/generated/seaborn.scatterplot.html)

```python
yr = 1997
filter = gapminder_df['year'] == yr
temp_df = gapminder_df[filter]
sns.scatterplot(
    data = temp_df,
    x = "gdpPercap",
    y = "lifeExp",
    # color="red",
    hue = "continent",
    alpha = 0.7,
    # s=60,
    size = "pop",
    sizes = (20, 600)
    )
plt.xscale('log')
```

## Creating Small Multiples

There are two ways we can create multiple mini scatterplots using Seaborn:

1. Use the Seaborn function **FacetGrid** with **scatterplot**. This method is more verbose but allows more customization.
2. Use the Seaborn function **relplot**. This option is simpler.

```python
g = sns.FacetGrid(
    data = gapminder_df,
    col="year",
    col_wrap = 3,
    hue = "continent"
    )
g.map(sns.scatterplot, "gdpPercap", "lifeExp")
plt.xscale('log')
```

```python
sns.relplot(
    data = gapminder_df,
    x = "gdpPercap",
    y = "lifeExp",
    col = "year",
    col_wrap = 3,
    hue = "continent",
    size = "pop",
    sizes = (40, 1000),
    alpha = 0.7,
    )
plt.xscale('log')
```

### IIc. Creating Animated Graphs with Plotly

25. An even better way to add time as a fourth dimension is by creating an animated graphic. We can do that using the [Plotly library](https://plotly.com/python/). Below, however, we will use [Plotly Express](https://plotly.com/python/plotly-express/), which is a simplified, easier-to-use version of Plotly.

https://towardsdatascience.com/recreating-gapminder-animation-in-2-lines-of-python-with-plotly-express-2060c73bedec

```python
import plotly.express as px
```

```python
fig = px.scatter(
    gapminder_df,
    x="gdpPercap",
    y="lifeExp",
    animation_frame="year",
    # animation_group="country",
    log_x = True,
    color="continent",
    range_x=[100,100000],
    range_y=[25,90],
    size="pop",
    size_max=45,
    hover_name="country",
    )

# fig.layout.updatemenus[0].buttons[0].args[1]["frame"]["duration"] = 300
fig.layout.updatemenus[0].buttons[0].args[1]["transition"]["duration"] = 8000
fig.show()
```
